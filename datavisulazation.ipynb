```python
import os
import pandas as pd
import matplotlib.pyplot as plt

df = None
current_plot = None


# =========================
# MAIN MENU
# =========================
def show_main_menu():
    print("\n" + "=" * 60)
    print("========== Data Analysis & Visualization Project ==========")
    print("Please select an option:")
    print("1. Load Dataset")
    print("2. Explore Data")
    print("3. Perform DataFrame Operations")
    print("4. Handle Missing Data")
    print("5. Generate Descriptive Statistics")
    print("6. Data Visualization")
    print("7. Save Visualization")
    print("8. Exit")
    print("=" * 60)


# =========================
# CHECK DATASET
# =========================
def require_data():
    if df is None:
        print("\nPlease load the dataset first!")
        return False
    return True


# =========================
# 1. LOAD DATASET
# =========================
def load_dataset():
    global df

    print("\n== Load Dataset ==")

    path = input(
        "Enter the path of the dataset (CSV file): "
    ).strip()

    try:
        df = pd.read_csv(path)

        print("Dataset loaded successfully!")
        print(f"Total records: {len(df)}")

    except FileNotFoundError:
        print("Dataset file not found!")

    except Exception as e:
        print("Error loading dataset:", e)


# =========================
# 2. EXPLORE DATA
# =========================
def explore_data():

    if not require_data():
        return

    print("\n== Explore Data ==")

    print("1. Display the first 5 rows")
    print("2. Display the last 5 rows")
    print("3. Display column names")
    print("4. Display data types")
    print("5. Display basic info")

    choice = input("Enter your choice: ")

    if choice == "1":

        print("\nFirst 5 rows:")
        print(df.head())

    elif choice == "2":

        print("\nLast 5 rows:")
        print(df.tail())

    elif choice == "3":

        print("\nColumn names:")
        print(list(df.columns))

    elif choice == "4":

        print("\nData types:")
        print(df.dtypes)

    elif choice == "5":

        print("\nBasic information:")
        df.info()

    else:
        print("Invalid choice!")


# =========================
# 3. DATAFRAME OPERATIONS
# =========================
def dataframe_operations():

    global df

    if not require_data():
        return

    print("\n== Perform DataFrame Operations ==")

    print("1. Select a column")
    print("2. Filter rows")
    print("3. Sort by a column")
    print("4. Add a new column")
    print("5. Delete a column")
    print("6. Display DataFrame shape")
    print("7. Display unique values")

    choice = input("Enter your choice: ")

    # Select column
    if choice == "1":

        column = input("Enter column name: ")

        if column in df.columns:
            print(df[column])
        else:
            print("Invalid column name!")

    # Filter rows
    elif choice == "2":

        column = input("Enter column name: ")
        value = input("Enter value to filter: ")

        if column in df.columns:

            result = df[
                df[column].astype(str).str.lower()
                == value.lower()
            ]

            print(result)

        else:
            print("Invalid column name!")

    # Sort
    elif choice == "3":

        column = input("Enter column name: ")

        if column in df.columns:

            order = input(
                "Sort ascending? (y/n): "
            ).lower()

            ascending = order != "n"

            print(
                df.sort_values(
                    by=column,
                    ascending=ascending
                )
            )

        else:
            print("Invalid column name!")

    # Add column
    elif choice == "4":

        column = input("Enter new column name: ")
        value = input("Enter value for the new column: ")

        df[column] = value

        print("New column added successfully!")
        print(df.head())

    # Delete column
    elif choice == "5":

        column = input("Enter column name to delete: ")

        if column in df.columns:

            df.drop(
                columns=[column],
                inplace=True
            )

            print("Column deleted successfully!")

        else:
            print("Invalid column name!")

    # Shape
    elif choice == "6":

        print(
            f"Rows: {df.shape[0]}, "
            f"Columns: {df.shape[1]}"
        )

    # Unique values
    elif choice == "7":

        column = input("Enter column name: ")

        if column in df.columns:

            print(df[column].unique())

        else:
            print("Invalid column name!")

    else:
        print("Invalid choice!")


# =========================
# 4. HANDLE MISSING DATA
# =========================
def handle_missing_data():

    global df

    if not require_data():
        return

    print("\n== Handle Missing Data ==")

    print("1. Display rows with missing values")
    print("2. Fill missing values with mean")
    print("3. Drop rows with missing values")
    print("4. Replace missing values with a specific value")

    choice = input("Enter your choice: ")

    # Check missing values
    if not df.isnull().values.any():

        print(
            "\nNo missing values found in the dataset!"
        )

        return

    if choice == "1":

        print("\nRows with missing values:")

        print(
            df[
                df.isnull().any(axis=1)
            ]
        )

    elif choice == "2":

        numeric_columns = (
            df.select_dtypes(
                include="number"
            ).columns
        )

        for column in numeric_columns:

            df[column] = df[column].fillna(
                df[column].mean()
            )

        print(
            "Missing numeric values "
            "filled with mean successfully!"
        )

    elif choice == "3":

        df.dropna(inplace=True)

        print(
            "Rows with missing values "
            "dropped successfully!"
        )

    elif choice == "4":

        value = input(
            "Enter the value to replace "
            "missing values: "
        )

        df.fillna(value, inplace=True)

        print(
            "Missing values replaced successfully!"
        )

    else:
        print("Invalid choice!")


# =========================
# 5. DESCRIPTIVE STATISTICS
# =========================
def descriptive_statistics():

    if not require_data():
        return

    print(
        "\n== Generate Descriptive Statistics =="
    )

    print("1. Summary statistics")
    print("2. Mean")
    print("3. Median")
    print("4. Minimum and Maximum")
    print("5. Standard deviation")

    choice = input("Enter your choice: ")

    numeric = df.select_dtypes(
        include="number"
    )

    if choice == "1":

        print(df.describe())

    elif choice == "2":

        print("\nMean:")
        print(numeric.mean())

    elif choice == "3":

        print("\nMedian:")
        print(numeric.median())

    elif choice == "4":

        print("\nMinimum:")
        print(numeric.min())

        print("\nMaximum:")
        print(numeric.max())

    elif choice == "5":

        print("\nStandard deviation:")
        print(numeric.std())

    else:
        print("Invalid choice!")


# =========================
# COLUMN FUNCTION
# =========================
def get_column(message):

    column = input(message)

    if column not in df.columns:

        print("Invalid column name!")

        return None

    return column


# =========================
# 6.1 BAR PLOT
# =========================
def bar_plot():

    global current_plot

    print("\n== Bar Plot ==")

    x = get_column(
        "Enter x-axis column name: "
    )

    y = get_column(
        "Enter y-axis column name: "
    )

    if not x or not y:
        return

    print("Generating bar plot...")

    grouped = df.groupby(x)[y].sum()

    plt.figure(figsize=(8, 5))

    grouped.plot(kind="bar")

    plt.xlabel(x)
    plt.ylabel(y)

    plt.title("Bar Plot")

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Bar plot displayed successfully!"
    )


# =========================
# 6.2 LINE PLOT
# =========================
def line_plot():

    global current_plot

    print("\n== Line Plot ==")

    x = get_column(
        "Enter x-axis column name: "
    )

    y = get_column(
        "Enter y-axis column name: "
    )

    if not x or not y:
        return

    print("Generating line plot...")

    plt.figure(figsize=(8, 5))

    plt.plot(
        df[x],
        df[y],
        marker="o"
    )

    plt.xlabel(x)
    plt.ylabel(y)

    plt.title("Line Plot")

    plt.xticks(rotation=30)

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Line plot displayed successfully!"
    )


# =========================
# 6.3 SCATTER PLOT
# =========================
def scatter_plot():

    global current_plot

    print("\n== Scatter Plot ==")

    x = get_column(
        "Enter x-axis column name: "
    )

    y = get_column(
        "Enter y-axis column name: "
    )

    if not x or not y:
        return

    print("Generating scatter plot...")

    plt.figure(figsize=(8, 5))

    plt.scatter(
        df[x],
        df[y]
    )

    plt.xlabel(x)
    plt.ylabel(y)

    plt.title("Scatter Plot")

    plt.grid(True)

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Scatter plot displayed successfully!"
    )


# =========================
# 6.4 PIE CHART
# =========================
def pie_chart():

    global current_plot

    print("\n== Pie Chart ==")

    label = get_column(
        "Enter label column name: "
    )

    value = get_column(
        "Enter value column name: "
    )

    if not label or not value:
        return

    grouped = df.groupby(
        label
    )[value].sum()

    print("Generating pie chart...")

    plt.figure(figsize=(7, 7))

    plt.pie(
        grouped.values,
        labels=grouped.index,
        autopct="%1.1f%%"
    )

    plt.title("Pie Chart")

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Pie chart displayed successfully!"
    )


# =========================
# 6.5 HISTOGRAM
# =========================
def histogram():

    global current_plot

    print("\n== Histogram ==")

    column = get_column(
        "Enter column name: "
    )

    if not column:
        return

    if not pd.api.types.is_numeric_dtype(
        df[column]
    ):

        print(
            "Histogram requires "
            "a numeric column!"
        )

        return

    print("Generating histogram...")

    plt.figure(figsize=(8, 5))

    plt.hist(
        df[column].dropna(),
        bins=10,
        edgecolor="black"
    )

    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.title("Histogram")

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Histogram displayed successfully!"
    )


# =========================
# 6.6 STACK PLOT
# =========================
def stack_plot():

    global current_plot

    print("\n== Stack Plot ==")

    x = get_column(
        "Enter x-axis column name: "
    )

    y1 = get_column(
        "Enter first y-axis column name: "
    )

    y2 = get_column(
        "Enter second y-axis column name: "
    )

    if not x or not y1 or not y2:
        return

    if not all(
        pd.api.types.is_numeric_dtype(
            df[column]
        )
        for column in [y1, y2]
    ):

        print(
            "Stack plot requires "
            "numeric y-axis columns!"
        )

        return

    print("Generating stack plot...")

    plt.figure(figsize=(8, 5))

    plt.stackplot(
        df[x],
        df[y1],
        df[y2],
        labels=[y1, y2]
    )

    plt.xlabel(x)
    plt.ylabel("Values")

    plt.title("Stack Plot")

    plt.legend()

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Stack plot displayed successfully!"
    )


# =========================
# VISUALIZATION MENU
# =========================
def data_visualization():

    if not require_data():
        return

    print("\n== Data Visualization ==")

    print("1. Bar Plot")
    print("2. Line Plot")
    print("3. Scatter Plot")
    print("4. Pie Chart")
    print("5. Histogram")
    print("6. Stack Plot")

    choice = input("Enter your choice: ")

    if choice == "1":
        bar_plot()

    elif choice == "2":
        line_plot()

    elif choice == "3":
        scatter_plot()

    elif choice == "4":
        pie_chart()

    elif choice == "5":
        histogram()

    elif choice == "6":
        stack_plot()

    else:
        print("Invalid choice!")


# =========================
# 7. SAVE VISUALIZATION
# =========================
def save_visualization():

    if current_plot is None:

        print(
            "\nNo visualization is available "
            "to save."
        )

        print(
            "Generate a plot first using option 6."
        )

        return

    print("\n== Save Visualization ==")

    filename = input(
        "Enter file name to save the plot "
        "(e.g., scatter_plot.png): "
    ).strip()

    if not filename:

        print("File name cannot be empty.")

        return

    os.makedirs(
        "output",
        exist_ok=True
    )

    if not os.path.splitext(filename)[1]:

        filename += ".png"

    path = os.path.join(
        "output",
        filename
    )

    try:

        current_plot.savefig(
            path,
            dpi=300,
            bbox_inches="tight"
        )

        print(
            f"Visualization saved as "
            f"{path} successfully!"
        )

    except Exception as e:

        print(
            "Error saving visualization:",
            e
        )


# =========================
# MAIN PROGRAM
# =========================
def main():

    while True:

        show_main_menu()

        choice = input(
            "Enter your choice: "
        )

        if choice == "1":

            load_dataset()

        elif choice == "2":

            explore_data()

        elif choice == "3":

            dataframe_operations()

        elif choice == "4":

            handle_missing_data()

        elif choice == "5":

            descriptive_statistics()

        elif choice == "6":

            data_visualization()

        elif choice == "7":

            save_visualization()

        elif choice == "8":

            print(
                "\nExiting the program. Goodbye!"
            )

            break

        else:

            print(
                "Invalid choice! "
                "Please try again."
            )


# =========================
# START PROGRAM
# =========================
if __name__ == "__main__":
    main()
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

df = None
current_plot = None


# =========================
# MAIN MENU
# =========================
def show_main_menu():
    print("\n" + "=" * 60)
    print("========== Data Analysis & Visualization Project ==========")
    print("Please select an option:")
    print("1. Load Dataset")
    print("2. Explore Data")
    print("3. Perform DataFrame Operations")
    print("4. Handle Missing Data")
    print("5. Generate Descriptive Statistics")
    print("6. Data Visualization")
    print("7. Save Visualization")
    print("8. Exit")
    print("=" * 60)


# =========================
# CHECK DATASET
# =========================
def require_data():
    if df is None:
        print("\nPlease load the dataset first!")
        return False
    return True


# =========================
# 1. LOAD DATASET
# =========================
def load_dataset():
    global df

    print("\n== Load Dataset ==")

    path = input(
        "Enter the path of the dataset (CSV file): "
    ).strip()

    try:
        df = pd.read_csv(path)

        print("Dataset loaded successfully!")
        print(f"Total records: {len(df)}")

    except FileNotFoundError:
        print("Dataset file not found!")

    except Exception as e:
        print("Error loading dataset:", e)


# =========================
# 2. EXPLORE DATA
# =========================
def explore_data():

    if not require_data():
        return

    print("\n== Explore Data ==")

    print("1. Display the first 5 rows")
    print("2. Display the last 5 rows")
    print("3. Display column names")
    print("4. Display data types")
    print("5. Display basic info")

    choice = input("Enter your choice: ")

    if choice == "1":

        print("\nFirst 5 rows:")
        print(df.head())

    elif choice == "2":

        print("\nLast 5 rows:")
        print(df.tail())

    elif choice == "3":

        print("\nColumn names:")
        print(list(df.columns))

    elif choice == "4":

        print("\nData types:")
        print(df.dtypes)

    elif choice == "5":

        print("\nBasic information:")
        df.info()

    else:
        print("Invalid choice!")


# =========================
# 3. DATAFRAME OPERATIONS
# =========================
def dataframe_operations():

    global df

    if not require_data():
        return

    print("\n== Perform DataFrame Operations ==")

    print("1. Select a column")
    print("2. Filter rows")
    print("3. Sort by a column")
    print("4. Add a new column")
    print("5. Delete a column")
    print("6. Display DataFrame shape")
    print("7. Display unique values")

    choice = input("Enter your choice: ")

    # Select column
    if choice == "1":

        column = input("Enter column name: ")

        if column in df.columns:
            print(df[column])
        else:
            print("Invalid column name!")

    # Filter rows
    elif choice == "2":

        column = input("Enter column name: ")
        value = input("Enter value to filter: ")

        if column in df.columns:

            result = df[
                df[column].astype(str).str.lower()
                == value.lower()
            ]

            print(result)

        else:
            print("Invalid column name!")

    # Sort
    elif choice == "3":

        column = input("Enter column name: ")

        if column in df.columns:

            order = input(
                "Sort ascending? (y/n): "
            ).lower()

            ascending = order != "n"

            print(
                df.sort_values(
                    by=column,
                    ascending=ascending
                )
            )

        else:
            print("Invalid column name!")

    # Add column
    elif choice == "4":

        column = input("Enter new column name: ")
        value = input("Enter value for the new column: ")

        df[column] = value

        print("New column added successfully!")
        print(df.head())

    # Delete column
    elif choice == "5":

        column = input("Enter column name to delete: ")

        if column in df.columns:

            df.drop(
                columns=[column],
                inplace=True
            )

            print("Column deleted successfully!")

        else:
            print("Invalid column name!")

    # Shape
    elif choice == "6":

        print(
            f"Rows: {df.shape[0]}, "
            f"Columns: {df.shape[1]}"
        )

    # Unique values
    elif choice == "7":

        column = input("Enter column name: ")

        if column in df.columns:

            print(df[column].unique())

        else:
            print("Invalid column name!")

    else:
        print("Invalid choice!")


# =========================
# 4. HANDLE MISSING DATA
# =========================
def handle_missing_data():

    global df

    if not require_data():
        return

    print("\n== Handle Missing Data ==")

    print("1. Display rows with missing values")
    print("2. Fill missing values with mean")
    print("3. Drop rows with missing values")
    print("4. Replace missing values with a specific value")

    choice = input("Enter your choice: ")

    # Check missing values
    if not df.isnull().values.any():

        print(
            "\nNo missing values found in the dataset!"
        )

        return

    if choice == "1":

        print("\nRows with missing values:")

        print(
            df[
                df.isnull().any(axis=1)
            ]
        )

    elif choice == "2":

        numeric_columns = (
            df.select_dtypes(
                include="number"
            ).columns
        )

        for column in numeric_columns:

            df[column] = df[column].fillna(
                df[column].mean()
            )

        print(
            "Missing numeric values "
            "filled with mean successfully!"
        )

    elif choice == "3":

        df.dropna(inplace=True)

        print(
            "Rows with missing values "
            "dropped successfully!"
        )

    elif choice == "4":

        value = input(
            "Enter the value to replace "
            "missing values: "
        )

        df.fillna(value, inplace=True)

        print(
            "Missing values replaced successfully!"
        )

    else:
        print("Invalid choice!")


# =========================
# 5. DESCRIPTIVE STATISTICS
# =========================
def descriptive_statistics():

    if not require_data():
        return

    print(
        "\n== Generate Descriptive Statistics =="
    )

    print("1. Summary statistics")
    print("2. Mean")
    print("3. Median")
    print("4. Minimum and Maximum")
    print("5. Standard deviation")

    choice = input("Enter your choice: ")

    numeric = df.select_dtypes(
        include="number"
    )

    if choice == "1":

        print(df.describe())

    elif choice == "2":

        print("\nMean:")
        print(numeric.mean())

    elif choice == "3":

        print("\nMedian:")
        print(numeric.median())

    elif choice == "4":

        print("\nMinimum:")
        print(numeric.min())

        print("\nMaximum:")
        print(numeric.max())

    elif choice == "5":

        print("\nStandard deviation:")
        print(numeric.std())

    else:
        print("Invalid choice!")


# =========================
# COLUMN FUNCTION
# =========================
def get_column(message):

    column = input(message)

    if column not in df.columns:

        print("Invalid column name!")

        return None

    return column


# =========================
# 6.1 BAR PLOT
# =========================
def bar_plot():

    global current_plot

    print("\n== Bar Plot ==")

    x = get_column(
        "Enter x-axis column name: "
    )

    y = get_column(
        "Enter y-axis column name: "
    )

    if not x or not y:
        return

    print("Generating bar plot...")

    grouped = df.groupby(x)[y].sum()

    plt.figure(figsize=(8, 5))

    grouped.plot(kind="bar")

    plt.xlabel(x)
    plt.ylabel(y)

    plt.title("Bar Plot")

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Bar plot displayed successfully!"
    )


# =========================
# 6.2 LINE PLOT
# =========================
def line_plot():

    global current_plot

    print("\n== Line Plot ==")

    x = get_column(
        "Enter x-axis column name: "
    )

    y = get_column(
        "Enter y-axis column name: "
    )

    if not x or not y:
        return

    print("Generating line plot...")

    plt.figure(figsize=(8, 5))

    plt.plot(
        df[x],
        df[y],
        marker="o"
    )

    plt.xlabel(x)
    plt.ylabel(y)

    plt.title("Line Plot")

    plt.xticks(rotation=30)

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Line plot displayed successfully!"
    )


# =========================
# 6.3 SCATTER PLOT
# =========================
def scatter_plot():

    global current_plot

    print("\n== Scatter Plot ==")

    x = get_column(
        "Enter x-axis column name: "
    )

    y = get_column(
        "Enter y-axis column name: "
    )

    if not x or not y:
        return

    print("Generating scatter plot...")

    plt.figure(figsize=(8, 5))

    plt.scatter(
        df[x],
        df[y]
    )

    plt.xlabel(x)
    plt.ylabel(y)

    plt.title("Scatter Plot")

    plt.grid(True)

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Scatter plot displayed successfully!"
    )


# =========================
# 6.4 PIE CHART
# =========================
def pie_chart():

    global current_plot

    print("\n== Pie Chart ==")

    label = get_column(
        "Enter label column name: "
    )

    value = get_column(
        "Enter value column name: "
    )

    if not label or not value:
        return

    grouped = df.groupby(
        label
    )[value].sum()

    print("Generating pie chart...")

    plt.figure(figsize=(7, 7))

    plt.pie(
        grouped.values,
        labels=grouped.index,
        autopct="%1.1f%%"
    )

    plt.title("Pie Chart")

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Pie chart displayed successfully!"
    )


# =========================
# 6.5 HISTOGRAM
# =========================
def histogram():

    global current_plot

    print("\n== Histogram ==")

    column = get_column(
        "Enter column name: "
    )

    if not column:
        return

    if not pd.api.types.is_numeric_dtype(
        df[column]
    ):

        print(
            "Histogram requires "
            "a numeric column!"
        )

        return

    print("Generating histogram...")

    plt.figure(figsize=(8, 5))

    plt.hist(
        df[column].dropna(),
        bins=10,
        edgecolor="black"
    )

    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.title("Histogram")

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Histogram displayed successfully!"
    )


# =========================
# 6.6 STACK PLOT
# =========================
def stack_plot():

    global current_plot

    print("\n== Stack Plot ==")

    x = get_column(
        "Enter x-axis column name: "
    )

    y1 = get_column(
        "Enter first y-axis column name: "
    )

    y2 = get_column(
        "Enter second y-axis column name: "
    )

    if not x or not y1 or not y2:
        return

    if not all(
        pd.api.types.is_numeric_dtype(
            df[column]
        )
        for column in [y1, y2]
    ):

        print(
            "Stack plot requires "
            "numeric y-axis columns!"
        )

        return

    print("Generating stack plot...")

    plt.figure(figsize=(8, 5))

    plt.stackplot(
        df[x],
        df[y1],
        df[y2],
        labels=[y1, y2]
    )

    plt.xlabel(x)
    plt.ylabel("Values")

    plt.title("Stack Plot")

    plt.legend()

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Stack plot displayed successfully!"
    )


# =========================
# VISUALIZATION MENU
# =========================
def data_visualization():

    if not require_data():
        return

    print("\n== Data Visualization ==")

    print("1. Bar Plot")
    print("2. Line Plot")
    print("3. Scatter Plot")
    print("4. Pie Chart")
    print("5. Histogram")
    print("6. Stack Plot")

    choice = input("Enter your choice: ")

    if choice == "1":
        bar_plot()

    elif choice == "2":
        line_plot()

    elif choice == "3":
        scatter_plot()

    elif choice == "4":
        pie_chart()

    elif choice == "5":
        histogram()

    elif choice == "6":
        stack_plot()

    else:
        print("Invalid choice!")


# =========================
# 7. SAVE VISUALIZATION
# =========================
def save_visualization():

    if current_plot is None:

        print(
            "\nNo visualization is available "
            "to save."
        )

        print(
            "Generate a plot first using option 6."
        )

        return

    print("\n== Save Visualization ==")

    filename = input(
        "Enter file name to save the plot "
        "(e.g., scatter_plot.png): "
    ).strip()

    if not filename:

        print("File name cannot be empty.")

        return

    os.makedirs(
        "output",
        exist_ok=True
    )

    if not os.path.splitext(filename)[1]:

        filename += ".png"

    path = os.path.join(
        "output",
        filename
    )

    try:

        current_plot.savefig(
            path,
            dpi=300,
            bbox_inches="tight"
        )

        print(
            f"Visualization saved as "
            f"{path} successfully!"
        )

    except Exception as e:

        print(
            "Error saving visualization:",
            e
        )


# =========================
# MAIN PROGRAM
# =========================
def main():

    while True:

        show_main_menu()

        choice = input(
            "Enter your choice: "
        )

        if choice == "1":

            load_dataset()

        elif choice == "2":

            explore_data()

        elif choice == "3":

            dataframe_operations()

        elif choice == "4":

            handle_missing_data()

        elif choice == "5":

            descriptive_statistics()

        elif choice == "6":

            data_visualization()

        elif choice == "7":

            save_visualization()

        elif choice == "8":

            print(
                "\nExiting the program. Goodbye!"
            )

            break

        else:

            print(
                "Invalid choice! "
                "Please try again."
            )


# =========================
# START PROGRAM
# =========================
if __name__ == "__main__":
    main()
```
